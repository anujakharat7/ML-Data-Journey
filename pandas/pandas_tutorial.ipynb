```python
s = pd.Series([1, 2, 3, 4, 5])
print(s)
print(type(s))

#index based access
print(s[2])
print(s[0])
```

```python
#labels
s = pd.Series([23, 24, 25], index = ["Adam", "Steve", "Bob"])
print(s)
print(s["Steve"])
print(s[1])
```

```python
#Series are homogeneous
s1 = pd.Series([1, 2, 3, "abc"])
print(s1)

#Series perform vectorized operation
s01 = pd.Series([1, 2, 3, 4])
s2 = pd.Series([10, 20, 30, 40])
print(s01 + s2)

#Series have mutable values but immutable size
print(s1[0])
s01[0] = 100
print(s01)
changed_s01 = s01.drop(0)
print(s01)
print(changed_s01)
```
